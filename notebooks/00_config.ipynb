# Central configuration notebook for the Insurance Claims Lakehouse project.

```python
def get_config(env="dev"):
    configs = {
        "dev": {
            "catalog_name": "insurance_lakehouse_dev",
            "raw_schema": "raw",
            "bronze_schema": "bronze",
            "silver_schema": "silver",
            "gold_schema": "gold",
            "governance_schema": "governance",
            "quarantine_schema": "quarantine",
            "source_volume": "insurance_source_files",
            "batch_id": "BATCH_001"
        },
        "test": {
            "catalog_name": "insurance_lakehouse_test",
            "raw_schema": "raw",
            "bronze_schema": "bronze",
            "silver_schema": "silver",
            "gold_schema": "gold",
            "governance_schema": "governance",
            "quarantine_schema": "quarantine",
            "source_volume": "insurance_source_files",
            "batch_id": "BATCH_001"
        },
        "prod": {
            "catalog_name": "insurance_lakehouse_prod",
            "raw_schema": "raw",
            "bronze_schema": "bronze",
            "silver_schema": "silver",
            "gold_schema": "gold",
            "governance_schema": "governance",
            "quarantine_schema": "quarantine",
            "source_volume": "insurance_source_files",
            "batch_id": "BATCH_001"
        }
    }

    if env not in configs:
        raise ValueError(f"Invalid env: {env}. Valid values are: {list(configs.keys())}")

    return configs[env]
```

```python
def build_table_name(config, schema_key, table_name):
    return f"{config['catalog_name']}.{config[schema_key]}.{table_name}"
```

```python
def build_volume_path(config):
    return f"/Volumes/{config['catalog_name']}/{config['raw_schema']}/{config['source_volume']}"
```

```python
def print_config(config):
    print("Project configuration")
    print("---------------------")
    print(f"Catalog:           {config['catalog_name']}")
    print(f"Raw schema:        {config['raw_schema']}")
    print(f"Bronze schema:     {config['bronze_schema']}")
    print(f"Silver schema:     {config['silver_schema']}")
    print(f"Gold schema:       {config['gold_schema']}")
    print(f"Governance schema: {config['governance_schema']}")
    print(f"Quarantine schema: {config['quarantine_schema']}")
    print(f"Source volume:     {config['source_volume']}")
    print(f"Batch ID:          {config['batch_id']}")
    print(f"Source path:       {build_volume_path(config)}")
```
